## Robot

```python


class Robot:
    @property
    def has_ball(self) -> bool:...
    @property
    def distance_from_ball(self) -> float:...
    @property
    def angle_to_ball(self) -> float:...

    @property
    def teammate_has_ball(self) -> bool:...
    @property
    def teammate_protocol(self) -> Protocol:...
    @property
    def teammate_start_pos(self) -> StartingPosition:...
    @property
    def teammate_distance_from_ball(self) -> float:...
    def __init__(self, starting_position:StartingPosition, team_colour:Colour):
        self.starting_position = starting_position
        self.team_colour = team_colour
        self.current_position = (self.starting_position % 2, self.starting_position/3)
        self.current_protocol = None


    def protocolSelector(self) -> Protocol:
        team_start_pos = self.teammate_start_pos
        team_protocol = self.teammate_protocol
        if self.current_protocol == None:
            match self.teammate_protocol:
                case Protocol.DEFENCE:
                    return Protocol.ATTACK
                case Protocol.ATTACK:
                    return Protocol.ATTACK
                case Protocol.INTERFERENCE:
                    return Protocol.ATTACK
                case _:
                    if self.starting_position == StartingPosition.FRONT_CENTRE:
                        return Protocol.ATTACK
                    elif team_start_pos == StartingPosition.FRONT_CENTRE:
                        return Protocol.INTERFERENCE
                    elif team_start_pos > self.starting_position:
                        return Protocol.INTERFERENCE
                    else:
                        return Protocol.ATTACK
        else:
            match self.current_protocol:
                case Protocol.ATTACK:
                    if any(self.has_ball, self.teammate_has_ball):
                        return Protocol.ATTACK
                    else:
                        if self.angle_to_ball > π/2 and self.angle_to_ball < 3*π/2:
                            return Protocol.DEFENCE
                        else: return Protocol.ATTACK
                case Protocol.DEFENCE:
                        if self.angle_to_ball < π/2 - π/6 and self.angle_to_ball < 3*π/2 + π/6:
                            return Protocol.ATTACK
                        else: return Protocol.DEFENCE
```

```python
def main():
    protocolSelector()
    print('a')

main()
```

## Enums

```python
class Protocol(Enum):
    DEFENCE = 0
    ATTACK = 1
    INTERFERENCE = 2

class Colour(Enum):
    YELLOW = 0
    BLUE = 1

class StartingPosition(Enum):
    BACK_LEFT = 0
    BACK_CENTRE = 1
    BACK_RIGHT = 2
    FRONT_LEFT = 3
    FRONT_CENTRE = 4
    FRONT_RIGHT = 5
```
